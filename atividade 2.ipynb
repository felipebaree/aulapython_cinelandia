```python
import pandas as pd
import numpy as np
```

```python
df_anp = pd.read_csv('ANP2025.csv', sep=';')    
```

```python
df_rj_etanol = df_anp.loc[(df_anp['Municipio'] == 'RIO DE JANEIRO') & (df_anp['Produto'] == 'ETANOL')].copy()
```

```python
df_rj_etanol = df_anp.loc[(df_anp['Municipio'] == 'RIO DE JANEIRO') & (df_anp['Produto'] == 'ETANOL')].copy()
```

```python
df_rj_etanol['Valor de Venda'] = (df_rj_etanol['Valor de Venda'].str.replace(',', '.', regex=False).astype(float))
```

```python
media_bairro = df_rj_etanol.groupby('Bairro')['Valor de Venda'].mean().reset_index()
```

```python
q3 = np.quantile(media_bairro['Valor de Venda'], 0.75)
```

```python
bairros_mais_caros = media_bairro.loc[media_bairro['Valor de Venda'] > q3]
```

```python
bairros_mais_caros.sort_values(by='Valor de Venda', ascending=False)
```
